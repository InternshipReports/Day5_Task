```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set(style="whitegrid")
%matplotlib inline
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
print("Basic Info:\n")
print(df.info())
```

```python
print("\n\nSummary Statistics:\n")
print(df.describe())
```

```python
print("\n\nMissing Values:\n")
print(df.isnull().sum())
```

```python
print("\n\nValue Counts for 'Sex':")
print(df['Sex'].value_counts())
```

```python
print("\n\nValue Counts for 'Pclass':")
print(df['Pclass'].value_counts())
```

```python
# Histogram for Age
plt.figure(figsize=(8, 5))
sns.histplot(df['Age'].dropna(), bins=30, kde=True)
plt.title('Age Distribution')
plt.show()
```

```python
sns.countplot(x='Survived', data=df)
plt.title('Survival Count')
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.show()
```

```python
# Countplot by Gender
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Survival by Gender')
plt.show()
```

```python
# Boxplot of Age vs Pclass
sns.boxplot(x='Pclass', y='Age', data=df)
plt.title('Age Distribution by Passenger Class')
plt.show()
```

```python
df.corr(numeric_only=True)
```

```python
sns.pairplot(df[['Survived', 'Age', 'Fare', 'Pclass']].dropna(), hue='Survived', kind='kde')
```

```python
sns.countplot(x='Pclass', data=df)
plt.title('Passenger Class Distribution')
plt.show()
```

```python
sns.barplot(x='Sex', y='Survived', data=df)
plt.title('Survival Rate by Gender')
plt.show()
```

```python
print("\nObservations:")
print("""
- Most passengers were in 3rd class and male.
- Females had a higher survival rate than males.
- Younger passengers had slightly higher chances of survival.
- Higher fare and 1st class were positively associated with survival.
""")
```
